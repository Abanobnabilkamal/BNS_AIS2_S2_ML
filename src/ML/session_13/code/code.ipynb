```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4])
y = np.array([2,2.8,3.6,4.5])
n = len(x)
w = 0
b = 0
```

```python

# Heperparameters 
learning_rate = 0.01
epochs = 1000
cost_value = []
print(f" inital W : {w}"
      f" inital b : {b}"
      f" learning rate : {learning_rate}"
      f" epochs : {epochs}")
print("training............")

```

```python
# Gradient Desent
for i in range(epochs):
    # prediction
    y_hat = w * x + b
    
    # error
    error = y_hat - y
    
    # Gradients
    d_w = np.mean(error * x)
    d_b = np.mean(error)
    
    # update parameters
    w = w - learning_rate * d_w
    b = b - learning_rate * d_b
    
    # prediction
    y_hat = w * x + b 
    
    # cost function 
    cost = 0.5 * np.mean((y_hat - y)**2 )
    cost_value.append(cost)
    if (i+1) % 100 == 0 or i == 0 :
        print(f"W : {w}"
        f" b : {b}"
        f" cost : {cost}"
        f" epochs : {i +1}")
```

```python
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(x, w * x +b, 'r' )
plt.title("Linear Regression")
plt.xlabel("x")
plt.ylabel("y")
plt.grid()

plt.subplot(1,2,2)
plt.plot(cost_value)
plt.title("cost vs Epochs")
plt.xlabel("Epoches")
plt.ylabel("Cost")
```
