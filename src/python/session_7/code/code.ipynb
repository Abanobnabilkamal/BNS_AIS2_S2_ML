```python
class shape:
    def calculate_area(self):
        pass
    
    
class circle(shape) :
    def __init__(self, radius) -> None:
        self.radius = radius

    def calculate_area(self):
        return 3.14* ((self.radius)**2)


class square(shape):
    def __init__(self,side) -> None:
            self.side = side
            
    def calculate_area(self):
        return (self.side*4)

x = circle(5)
y = square(4)
print("Area: ", x.calculate_area())
print("Area: ", y.calculate_area())
```

```python
class MOdel:
    def predict(self, x):
        raise NotImplementedError

class LinearRegression(MOdel):
    def predict(self, x):
        return " Linear Regression prediction"
class DecisionTree (MOdel) :
    def predict(self, x):
        return " DecisionTree prediction"


models = [LinearRegression(),DecisionTree()]
for model in models:
    print(model.predict(None))
```
