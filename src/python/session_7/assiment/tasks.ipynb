```python
# the function

def bank_account(initial_balance):
    balance =initial_balance
    def deposit(amount):
        nonlocal balance
        if amount < 0 :
            print("Deposit amount cannot be negative.")
        else :
            balance+=amount
            print(f"Deposited {amount}. New balance: {balance}")
            
    def withdraw(amount):
        nonlocal balance
        if amount < 0 or amount > balance:
            print("withdraw amount cannot be negative.")
        else:
            balance -= amount
            print(f"withdraw {amount}. New balance: {balance}")    
        
    def check_balance():
        print(f"Current balance: {balance}")
        
    return deposit,withdraw,check_balance

deposit,withdraw,check_balance=bank_account(1000)
deposit(100)
withdraw(500)
check_balance()
```

```python
# the class
class BankAccount:
    def __init__(self,balance):
        self.balance=balance
    
    def deposit(self,amount):
        if amount <= 0:
            print("Deposit amount cannot be negative.") 
        else:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
            
    
    def withdraw(self,amount):
        if amount < 0 or amount > self.balance:
            print("withdraw amount cannot be negative.")
        else:
            self.balance -= amount
            print(f"withdraw {amount}. New balance: {self.balance}")
    
    @property
    def check_balance(self):
        print(f"Current balance: {self.balance}")
        
        
            
        
```

```python
bank_1=BankAccount(1000)

print(bank_1.withdraw(500))
print(bank_1.check_balance)
print(bank_1.withdraw(500))
```

```python
from abc import ABC,abstractmethod

class Animal(ABC):
    
    @abstractmethod
    def make_sound(self): 
        pass
    
    def describe(self):
        print("this is the main class")
        
class Cat(Animal):
    def make_sound(self): 
        return "nao nao"
    
    def describe(self):
        print("this is a cat \n")
        
        
class Dog(Animal):
    def make_sound(self): 
        return "ho ho"
        
    def describe(self):
        print("this is a dog \n") 
        
        
class Cow(Animal):
    def make_sound(self): 
        return "wa wa"
    
    def describe(self):
        print("this is a Cow \n") 

            
    
        
```

```python
cat_1=Cat()
dog_1=Dog()
cow_1=Cow()

print(cat_1.make_sound())
cat_1.describe()

print(dog_1.make_sound())
dog_1.describe()

print(cow_1.make_sound())
cow_1.describe()

```

```python
class TextFileReader:
    def __init__(self,file_path):
        self.file_path=file_path
    
    def read_file(self):
        with open(self.file_path,"r") as file:
            self.content = file.read()
            
    def count_lines(self):
        return len(self.content.splitlines())
    
    def count_words(self):
        return len(self.content.split())
    
    def count_characters(self):
        
        return len(self.content)
    
    def display_content(self):
        print(self.content)
        
reader = TextFileReader("abanob.txt")

reader.read_file()

print("Number of lines:", reader.count_lines())
print("Number of words:", reader.count_words())
print("Number of characters:", reader.count_characters())

print("\nFile Content:")
reader.display_content()
 
        
```

```python
words = ["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"]

def count(words):
    
    unique_words = set(words)
    counts = {word: 0 for word in unique_words}
    for i in unique_words:
        for j in words:
            if i == j:
                counts[i] += 1
    return counts

print(count(words))
```

```python
def read_txt_file(file_path):
    """Reads the contents of a text file and returns it as a string."""

    try:
        with open(file_path, "r", encoding="utf-8") as file:
            content = file.read()

        return content

    except FileNotFoundError:
        return f"Error: The file '{file_path}' was not found."

    except IOError:
        return "Error: An error occurred while reading the file."


class UserExtractor:

    def __init__(self):
        self.file_path = input("Enter the path of the file (.txt): ").strip()
        self.usernames = {}

    def extract_usernames(self):
        """Extracts usernames from the text file and stores them in a dictionary."""

        content = read_txt_file(self.file_path)

        if content.startswith("Error:"):
            return content

        the_UserAndPassword = content.splitlines()

        for line in the_UserAndPassword:

            if ":" in line:
                name = line[0:line.find(":")]
                password = line[line.find(":") + 1:]

                self.usernames[name] = password

        return self.usernames


if __name__ == "__main__":

    read_1 = UserExtractor()

    print(read_1.extract_usernames())

    print("*" * 20)

    print("File path:", read_1.file_path)
```
